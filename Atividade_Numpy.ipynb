Nível 2: Intermediário — Múltiplas Filiais e Falhas de Sistema
Agora você precisa compilar os dados de três filiais diferentes ao longo de quatro dias. A matriz resultante deverá representar as lojas nas linhas e os dias nas colunas.



Passo 1: Crie uma matriz 2D a partir das seguintes listas para representar as vendas: Loja A: [200, 220, np.nan, 250] (Houve uma falha no sistema no 3º dia, gerando um valor não-número). Loja B: [150, 180, 160, 190] Loja C: [300, 310, 290, 330]

```python
LojaA= [200, 220, np.nan, 250]
LojaB= [150, 180, 160, 190]
LojaC= [300, 310, 290, 330]

print(np.array([LojaA, LojaB, LojaC]))
```

 Passo 2: Calcule o total de peças vendidas por cada loja (agregando pelas linhas). Para isso, aplique a função de soma especificando o parâmetro axis=1.

```python
vendas_filiais = np.array([LojaA, LojaB, LojaC])
total_por_loja = np.sum(vendas_filiais, axis=1)
print('Total de peças vendidas por loja:', total_por_loja)
```

Passo 3: A diretoria quer saber a média diária geral da rede. Como os dados reais frequentemente possuem lacunas, calcule a média utilizando a função do NumPy que ignora os valores nan. Se você usar a média normal, o resultado será contaminado e invalidará o relatório.

```python
media_diaria_geral = np.nanmean(vendas_filiais)
print('Média diária geral da rede (ignorando NaNs):', media_diaria_geral)
```

 Passo 4: A meta diária de cada loja é vender 200 peças. Utilize a função np.where() para criar uma nova matriz condicional, onde os valores maiores ou iguais a 200 sejam substituídos pelo texto 'Meta Atingida', e os valores inferiores sejam substituídos por 'Abaixo'.

```python
metas_lojas = np.where(vendas_filiais >= 200, 'Meta Atingida', 'Abaixo')
print('Matriz condicional de metas das lojas:\n', metas_lojas)
```

Nível 3: Avançado — Categorização e Destaques de Marketing
Nesta etapa final, você fará uma análise focada no perfil das vendas e no sucesso das campanhas publicitárias.



Passo 1: Para simular uma base de clientes, utilize o gerador moderno com uma semente fixa (np.random.default_rng()) para que os resultados sejam reproduzíveis. Crie uma amostra com 50 vendas sorteadas entre as opções ['Eletrônicos', 'Roupas', 'Casa'].

```python
rng = np.random.default_rng(seed=42) # Usando uma semente fixa para reprodutibilidade
opcoes_venda = ['Eletrônicos', 'Roupas', 'Casa']
amostra_vendas = rng.choice(opcoes_venda, size=50)
print('Amostra de 50 vendas:\n', amostra_vendas)
```

Passo 2: Determine o volume de vendas de cada departamento. Utilize a função np.unique habilitando o argumento return_counts=True para obter os valores únicos e suas respectivas contagens — esta é uma forma rápida de fazer o equivalente numérico de contagem de categorias.

```python
departamentos, contagens = np.unique(amostra_vendas, return_counts=True)
print('Volume de vendas por departamento:')
for dep, count in zip(departamentos, contagens):
    print(f'- {dep}: {count} vendas')
```

 Passo 3: O time de marketing enviou um array com o faturamento de cinco campanhas simultâneas: [12000, 45000, 23000, 89000, 31000]. Identifique rapidamente a campanha campeã utilizando o método que retorna o índice (posição) do maior valor do array.

```python
faturamento_campanhas = np.array([12000, 45000, 23000, 89000, 31000])
indice_campea = np.argmax(faturamento_campanhas)
print(f'Faturamento das campanhas: {faturamento_campanhas}')
print(f'O índice da campanha campeã é: {indice_campea}')
print(f'O faturamento da campanha campeã é: {faturamento_campanhas[indice_campea]}')
```
